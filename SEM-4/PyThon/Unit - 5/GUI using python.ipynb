```python
from tkinter import *
```

# Basic GUI

```python
top = Tk()
top.title("Notepad")
top.minsize(width = 500,height = 500)
top.maxsize(width = 800,height = 800)

label = Label(top,text = "This is a label!!").pack(side = TOP)
top.mainloop()
```

# Grid Layout

```python
top = Tk()

top.title("DEMO GUI:")

#top.minsize(width = 300,height = 300)
#top.maxsize(width = 500,height = 500)

top.geometry('1090x800')
#to use geometry disable or comment out minsize and maxsize attribute

label = Label(top,text = "That's a Label:").grid(row = 0,column = 0)
txt = Entry(top,width = 15)
txt.grid(row = 0,column = 1)

label = Label(top,text = "What's on your Mind??").grid(row = 1,column = 0)
txt = Entry(top,width = 30)
txt.grid(row = 1,column = 1)

button = Button(top,text = "OK?").grid(row = 2,columnspan = 2)

top.mainloop()
```

# Using Place Instead of Grid 

```python
top = Tk()

top.title("DEMO GUI:")

#top.minsize(width = 300,height = 300)
#top.maxsize(width = 500,height = 500)

top.geometry('400x400')
#to use geometry disable or comment out minsize and maxsize attribute

label = Label(top,text = "That's a Label:").place(x = 0,y = 0)
txt = Entry(top,width = 15)
txt.place(x = 100,y = 0)

label = Label(top,text = "What's on your Mind??").place(x = 0,y = 30)
txt = Entry(top,width = 30)
txt.place(x = 140,y = 30)

button = Button(top,text = "OK?",width = 30).place(x = 0,y = 60)

top.mainloop()
```

# Making Checkboxes

```python
master = Tk()
master.minsize(width = 300,height = 200)
var1 = IntVar()
Checkbutton(master, text = 'Agree', variable = var1).grid(row = 0)
var2 = IntVar()
Checkbutton(master, text = 'Disagree',variable = var2).grid(row = 1)
mainloop()
```
