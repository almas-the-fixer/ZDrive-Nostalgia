```python
import tkinter as tk
from tkinter import messagebox
import mysql.connector



def save_to_database():
    name = entry_name.get()
    roll_number = entry_roll_number.get()
    marks1 = entry_marks1.get()
    marks2 = entry_marks2.get()
    marks3 = entry_marks3.get()

    try:
        conn = mysql.connector.connect(
            host="localhost",
            user="root",
            password="",
            database="myDB"
        )
        cursor = conn.cursor()
        cursor.execute('''CREATE TABLE IF NOT EXISTS students (
                        id INT PRIMARY KEY AUTO_INCREMENT,
                        name varchar(30) NOT NULL,
                        roll_number int NOT NULL,
                        marks1 float NOT NULL,
                        marks2 float NOT NULL,
                        marks3 float NOT NULL)''')
        cursor.execute("INSERT INTO students (name, roll_number, marks1, marks2, marks3) VALUES (%s, %s, %s, %s, %s)",
                       (name, roll_number, marks1, marks2, marks3))
        conn.commit()
        conn.close()
        root.destroy()
        messagebox.showinfo("Success", "Student information saved successfully.")
    except Exception as e:
        messagebox.showerror("Error", f"Error occurred: {e}")
def calculate_percentage():
    try:
        total_marks = float(entry_marks1.get()) + float(entry_marks2.get()) + float(entry_marks3.get())
        percentage = (total_marks / 300) * 100
        label_percentage.config(text="Percentage: {:.2f}%".format(percentage))
    except ValueError:
        messagebox.showerror("Error", "Please enter valid marks.")

# Create main window
root = tk.Tk()
root.title("Student Information Form")

# Create and pack labels and entry widgets for student information
label_name = tk.Label(root, text="Name:")
label_name.grid(row=0, column=0, padx=10, pady=5)
entry_name = tk.Entry(root)
entry_name.grid(row=0, column=1, padx=10, pady=5)

label_roll_number = tk.Label(root, text="Roll Number:")
label_roll_number.grid(row=1, column=0, padx=10, pady=5)
entry_roll_number = tk.Entry(root)
entry_roll_number.grid(row=1, column=1, padx=10, pady=5)

label_marks1 = tk.Label(root, text="Marks 1:")
label_marks1.grid(row=2, column=0, padx=10, pady=5)
entry_marks1 = tk.Entry(root)
entry_marks1.grid(row=2, column=1, padx=10, pady=5)

label_marks2 = tk.Label(root, text="Marks 2:")
label_marks2.grid(row=3, column=0, padx=10, pady=5)
entry_marks2 = tk.Entry(root)
entry_marks2.grid(row=3, column=1, padx=10, pady=5)

label_marks3 = tk.Label(root, text="Marks 3:")
label_marks3.grid(row=4, column=0, padx=10, pady=5)
entry_marks3 = tk.Entry(root)
entry_marks3.grid(row=4, column=1, padx=10, pady=5)

# Button to calculate percentage
button_calculate = tk.Button(root, text="Calculate Percentage", command=calculate_percentage)
button_calculate.grid(row=5, column=0, columnspan=2, pady=10)

# Label to display percentage
label_percentage = tk.Label(root, text="")
label_percentage.grid(row=6, column=0, columnspan=2, pady=5)

# Button to save student information
button_save = tk.Button(root, text="Save", command=save_to_database)
button_save.grid(row=7, column=0, columnspan=2, pady=10)

root.mainloop()
```
