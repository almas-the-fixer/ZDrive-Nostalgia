```python
import mysql.connector as sql
```

# Creating  and Connecting to DB

```python
mydb = sql.connect(
    host = "localhost",
    user = "root",
    password = ""
)

print(mydb)

cur = mydb.cursor()
cur.execute("CREATE DATABASE IF NOT EXISTS Demo")

cur.execute("USE Demo")
```

# Creating Table Employee

```python
cur.execute("CREATE TABLE EMP (EMP_ID INT PRIMARY KEY AUTO_INCREMENT,EMP_NAME VARCHAR(255),BASIC_PAY INT,DA INT,HRA INT)")
```

# Inserting values alt method

```python
def insert_records(mydb):
    try:
        for i in range(1,6):
            name = f"Employee {i}"
            basic = 5000 + (i * 1000)
            da = basic * 0.1
            hra = basic * 0.05
            cur.execute("INSERT INTO EMP (EMP_NAME,BASIC_PAY,DA,HRA) VALUES (%s,%s,%s,%s)",(name,basic,da,hra))
        mydb.commit()
        print("Records inserted successfully")
        cur.close()
    except mysql.connector.Error as err:
        print("Error: ",err)

insert_records(mydb)
```
