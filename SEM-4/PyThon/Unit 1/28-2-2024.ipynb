```python
print("Program to convert km to other units")

km = int(input("Enter KM value: "))
m = km * 1000
print("Entered value: ",km,"Kilometers is equal to: ",m,"meters")

cm = km * 100000
print("Entered value: ",km,"Kilometers is equal to: ",cm,"centimeters")

inch = km * 39370.1
print("Entered value: ",km,"Kilometers is equal to: ",inch,"inches")

feet = km * 3280.84
print("Entered value: ",km,"Kilometers is equal to: ",feet,"feet")
```

```python
print("Program to convert Celcius to farenheit")

celcius = int(input("Enter Value in celcius: "))

farenheit = (9/5)*celcius+32

print("Celciues value: ",celcius,"in farenheit is: ",farenheit,"farenheit")

f = int(input("Enter farenheit value: "))

c = 5/9 * (f-32)
print("Entered farenheit value is:",c,"celcius")
```

```python
print("Program to compute simple interest and compound interest")

principal = int(input("Enter Principal value: "))
interest = int(input("Enter Rate of Interest: "))
n = int(input("Enter Years: "))

si = (principal * interest * n)/100
print("Simple interest is: ",si)

#incomplete

ci = principal * (1 + (interest/100)**n) - principal
print("Compound interest is : ",ci)
```
