```python
print("Hello Day 2 of Python")
X = input("Enter a Number: ")
print(type(X))

length = len(X)

total = 0
for digit in X:
    ans = int(digit)**length
    total = total + ans
print(total)

if total == int(X):
    print("Number is Armstrong!")
else:
    print("Number is NOT Armstrong!")
```

```python
x = int(input("Enter a Number: "))
print("Your Entered Number:",x)
for i in range(2,x-1):
    if x%i == 0:
        print("NOT Prime Number!")
        break
else:
        print("Prime Number!")
```

```python
r = int(input("Enter a range: "))
print("Your Entered range:",r)

for i in range(1,r):
    for j in range(1,i-1):
        if i%j != 0:
            print("Prime Number: ",i)
```
