```python
import numpy as np
```

```python
x = np.array([[1,2],[3,4]])
y = np.array([[12,30]])
z = np.concatenate((x,y),axis=0) #keeps it 2d
z1 = np.concatenate((x,y),axis = None) # flattens array to 1d
z2 = np.concatenate((x,y.T),axis = 1)

print("------------Concatenate function------------------")
print(z)
print(z1)
print(z2)

app = np.append(x,y,axis=None)
app2 = np.append(x,y,axis=0)
app3 = np.append(x,y.T,axis=1)

print("----------------append function -----------------")
print(app)
print(app2)
print(app3)
```

```python
# Numpy Where Function

a = np.arange(12)

print("Array of 12 elements: ")
print(a)

print("Where element is greater then 6 multiply it by 5!")
b = np.where(a<6,a,5*a) # where(condition,action_if_true,action_if_false)
#it checks if element is less than 6 or not if it is'nt then it multiplies it by 5
print(b)

c = np.where([[True,False],[True,True]],[[1,2],[3,4]],[[9,8],[7,6]])

print("If True the it will Put a Value Corresponding to true's index from 1st matrix i.e true is at(0,0) if false it will take value from second matrix matching false's index!! Tricky")
print(c)
```

```python
#Numpy Random Module!!

x = np.random.randn(2,2)
y = np.random.randn()
print(x)
print(y)

#Generate Random integers giving limits
#np.random.randint(upperlimit,size = size of array)
n =  np.random.randint(3,size = 10)
print(n)
```

# MaTplot Library

```python
import matplotlib.pyplot as plt
```

```python
xcord = np.array([0,5])
ycord = np.array([0,8])

x = np.array([1,2,6,8])
y = np.array([3,8,1,0])

plt.plot(xcord,ycord,'_')
#plot(xcoordinate,ycoordinate,marker)
#we can use o,*,^,p,X etc for markers!!!
plt.show()

#if we dont enter y value it will by default use 0,1,2,3... values for y
#also if we write just X it won't draw a line but if we write marker = 'X' it will draw line and plot using X
#ms is marker size and mec is marker edge color
#linestyle changes line style i.e dotted dashed
plt.plot(x,y,marker="*",color='green',linestyle='-.',mec='y',mfc="red",ms = 20,linewidth="5")
plt.show()

plt.scatter(x,y)
plt.show()
```

```python
xx = np.array([10,20,30,40])
yy = np.array([50,40,45,25])

#plotting 2 lines
plt.plot(xx)
plt.plot(yy)

#create a dictionary for font properties
font1 = {'family':'Ink Free','color':'blue','size':20}
#Adding title and font properties using fontdict parameter 
plt.title("Name VS Number:",fontdict=font1)

plt.xlabel("Name")
plt.ylabel("number")
plt.show()
```

```python
#Plotting student grade and marks

student_marks = [50,60,70,80,90]
student_grade = ['B','B','B+','B+','A']

plt.xlabel = ("Student Marks")
#taking plt.axes in alias ax
ax = plt.axes()
#setting background color to black
ax.set_facecolor("Black")

plt.ylabel("Student Grade")

plt.title("Student Marks v\s Student Grade: ",loc = "right") # loc parameter is used to specify location of title legal values are left,right,center center is default
plt.plot(student_marks,student_grade)
```

# Bar Graph

```python
#The Bar graph can be drawn with bar() function
g = np.array(['Almas','Harsh','Nihar','Krish','Burhan'])
h = np.array([10,20,30,40,50])

plt.bar(g,h,color="magenta",width=0.5)#color and width can be applied also
plt.show()
plt.barh(g,h) #plots a horizontal bar graph nothing special
plt.show()
```

# Histogram

```python
#Histogram can be plotted using hist() function
x = np.random.normal(170,10,250)
plt.hist(x)
plt.show()
```

# Pie Chart

```python
#Pie Chart can be plotted using pie() function 
i = np.array([35,25,15,10])
mylabels = ["Apple","Orange","Banana","Dates"]
mycolors = ["Red","Orange","Yellow","Brown"]
myexplode = [0.2,0.4,0.6,0.8] #places specific or edges far away acc to values given
plt.pie(i,labels = mylabels,startangle=90,explode = myexplode,shadow = True,colors = mycolors) # we can specify start angle default is 0
plt.legend(title = 'Fruits') # can also place title in legend
# shadow = true places shadows
# colors takes a list of colors
# legend gets legend
plt.show()
#by default its edge starts from x axis and moves counterclockwise
```
