# 1D Array using numpy 

```python
import numpy as np

n = int(input("Enter the length of array: "))
ls = []
type(ls)
for i in range(0,n):
    ele = int(input("Enter an Element: "))
    ls.append(ele)
arr = np.array(ls)
print(arr)
type(arr)

#adding 1 to each element
for i in range(0,n):
    arr[i]+=1
print(arr)
```

# 2D matrix and their sum

```python
print("Warning!\nPython doesn't allow rectangular matrices so please use same value for rows and cols!")
row = int(input("Enter no of rows: "))
col = int(input("Enter no of cols: "))

l1 = []
l2 = []
l3 = []
l4 = []
l5 = []
l6 = []
#First matrix
for i in range(0,row):
    ele1 = int(input("Enter element: "))
    l1.append(ele1)
for i in range(0,col):
    ele2 = int(input("Enter Element: "))
    l2.append(ele2)

#Second Matrix
print("Second Matrix: ")
for i in range(0,row):
    e = int(input("Enter element: "))
    l3.append(e)
for i in range(0,col):
    f = int(input("Enter Element: "))
    l4.append(f)

#Adding 2
for i in range(0,row):
    l5.append(l1[i]+l3[i])
    l6.append(l2[i]+l4[i])
    
arr1 = np.array([l1,l2])
arr2 = np.array([l3,l4])
arr3 = np.array([l5,l6])

print(arr1)
print(arr2)
print("Sum of Both: ")
print(arr3)
type(arr1)
```

# Scalar Multiplation

```python
no = int(input("Enter no of cols: "))

li1 = []
li2 = []

for i in range(0,row):
    ele1 = int(input("Enter element: "))
    li1.append(ele1)
for i in range(0,col):
    ele2 = int(input("Enter Element: "))
    li2.append(ele2)

array1 = np.array([li1,li2])

print(array1)

number = int(input("Enter a number to multiply whole matrix with: "))

for i in range(0,no):
    array1[i]= array1[i] * number
print(array1)
```

```python
#multiplication of a matrix using numpy '@' or dot operator

print("Enter Rows and Cols for First Matrix: ")

r = int(input("Number of Rows: "))
c = int(input("Number of Cols: "))

print("Enter Rows and Cols for Second Matrix: ")

r2 = int(input("Number of Rows: "))
c2 = int(input("Number of Cols: "))

ar = []

print("Enter First Matrix Elements: ")
for i in range(r):
    row = []
    for j in range(c):
        element = int(input("Enter Element: "))
        row.append(element)
    ar.append(row)
demo = np.array(ar)        

ar2 = []

print("Enter Second Matrix Elements: ")
for i in range(r2):
    ro = []
    for j in range(c2):
        elem = int(input("Enter Element: "))
        ro.append(elem)
    ar2.append(ro)
demo2 = np.array(ar2) 

print("First Entered Matrix: ",demo)
print("Second Entered Matrix: ",demo2)

if (c==r2): # checking for compatibility ^_^
    result = np.dot(demo,demo2) 
    #we can also do this with '@'
    # result = demo @ demo2
    print("Multiplication of Both matrices is: ",result)
else:
    print("Multiplication not possible due to incompatibility!!")
```
