# Catching Spectifc Exception

```python
def fun(a):
    if a<4:
        b = a/(a-3)
    print("Value of b : ",b)
    
try:
#    fun(3)
    fun(5)
except ZeroDivisionError:
    print("ZeroDivisionError occured and Handled")
except NameError:
    print("NameError Occured and Handled")
```

# Try with Else Clause

```python
def AbyB(a , b):
    try:
        c = ((a+b)/(a-b))
    except ZeroDivisionError:
        print("a/b result in 0")
    else:
        print(c)

AbyB(2.0,3.0)
AbyB(3.0,3.0)
```

# Raise an Exception

```python
num = [1,2,3,4]

if len(num)>3:
    raise Exception("Length of given list must be under 3!")
```

# Finally Keyword

```python
try:
    div = 4 // 0
    print(div)
except ZeroDivisionError:
    print("Attempting to Divide by Zero!")
finally:
    print("This code is from finally keyword!")
```

```python
try:
    n1 = input("Enter a number: ")
    n2 = input("Enter second number: ")
    int(n1)
    int(n2)
except TypeError:
    print("Enter Numbers only!")
```
