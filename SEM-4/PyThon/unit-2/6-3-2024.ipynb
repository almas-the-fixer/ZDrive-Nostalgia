```python
class Rectangle:
    def __init__(self,length=None,width=None):
        if(length == None and width == None): #if no value is entered then it will take default values according to condition
            self.length = 5
            self.width = 10
        self.length = length
        self.width = width
    def area_pm(self):
        area = self.length * self.width
        perimeter = 2*(self.length+self.width)
        print("Area: ",area,"\nPerimeter: ",perimeter)
r1 = Rectangle(10,20)
r1.area_pm()
```

```python
class Account:
    def __init__(self,AccNo,Name,B_Code,B_Name,Balance):
        self.AccNo = AccNo
        self.Name = Name
        self.B_Code = B_Code
        self.B_Name = B_Name
        self.Balance = Balance
        
    def display(self):
        print("Account Number: ",self.AccNo)
        print("Name: ",self.Name)
        print("Branch Name: ",self.B_Name)
        print("Branch Code: ",self.B_Code)
        print("Balance: ",self.Balance)
        
    def deposit(self):
        amt = int(input("Enter Amount to Deposit: "))
        self.Balance += amt
        print("Updated Balance: ",self.Balance)

#------------------------------------------------------------

#a1 = Account(101,"abc",1001,"sarkhej",10000)
#a1.display()
print("------Displaying-------")
#a1.deposit()
print("------Displaying Again After Deposit-------")
#a1.display()

#------------------------------------------------------------

class SavingAccount(Account):
    def __init__(self,AccNo,Name,B_Code,B_Name,Balance):
        super().__init__(AccNo,Name,B_Code,B_Name,Balance)
    def withdraw(self):
        w_amt = int(input("Enter amount to withdraw: "))
        
        if w_amt > self.Balance:
            print("You Dont have sufficient Balance to Withdraw !!")
        else:
            self.Balance -= w_amt
            print("Amount Withdrawed Sucessfully!! Your current Balance is: ",self.Balance)
s1 = SavingAccount(102,"def",1002,"sanand",10000)
s1.withdraw()

class CurrentAccount(Account):
    def __init__(self,AccNo,Name,B_Code,B_Name,Balance):
        super().__init__(AccNo,Name,B_Code,B_Name,Balance)
    def withdraw(self):
        w_amt = int(input("Enter amount to withdraw: "))
        
        if w_amt > self.Balance:
            print("You Dont have sufficient Balance to Withdraw Overdrafting your account !!")
            self.Balance -= w_amt
            print("Amount Withdrawed Sucessfully using overdraft!! Your current Balance is: ",self.Balance)
        else:
            self.Balance -= w_amt
            print("Amount Withdrawed Sucessfully!! Your current Balance is: ",self.Balance)
c1 = CurrentAccount(103,"efg",1003,"Juhapura",10000)
c1.withdraw()
```

```python
class Person:
    def __init__(self,name=None,age=None):
        if(name == None or age == None):
            self.name = "Mahesh"
            self.age = 24
        else:
            self.name = name
            self.age = age
    def display(self):
        print("Name: ",self.name,"\nAge: ",self.age)
class Employee(Person):
    def __init__(self,name,age,role):
        super().__init__(name,age)
        self.role = role
    def E_Display(self):
        print("Name: ",self.name,"\nAge: ",self.age,"\nRole: ",self.role)
class PartTimeEmp(Employee):
    def __init__(self,name,age,role,hrs,rate_per_hour):
        super().__init__(name,age,role)
        self.hrs = hrs
        self.rate_per_hour = rate_per_hour
    def PE_Display(self):
        print("Name: ",self.name,"\nAge: ",self.age,"\nRole: ",self.role,"\nWorking Hours: ",self.hrs,"\nRate per Hour: ",self.rate_per_hour)
class FullTimeEmp(Employee):
    def __init__(self,name,age,role,da,hra):
        super().__init__(name,age,role)
        self.da = da
        self.hra = hra
    def FE_Display(self):
        print("Name: ",self.name,"\nAge: ",self.age,"\nRole: ",self.role,"\nDa: ",self.da,"\nHRA: ",self.hra)
#e1 = Employee("Almas",19,"Developer")
#e1.E_Display()

#p1 = PartTimeEmp("Almas",19,"Developer",6,500)
#p1.PE_Display()

f1 = FullTimeEmp("Almas",19,"Developer",5000,10000)
f1.FE_Display()
```
