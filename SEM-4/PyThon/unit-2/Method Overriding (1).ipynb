# Method Overriding

```python
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def Display(self):
        print("Name: ",self.name,"\nAge: ",self.age)
            class Employee(Person):
                def __init__(self,name,age,eid):
                    Person.__init__(self,name,age)
                    self.eid = eid

                def Display(self):
                    super().Display()    #If we comment out this function it will search for Display() in superclass!!
                    print("E_ID: ",self.eid)

class PartTimeEmployee(Employee):
    def __init__(self,name,age,eid,hrs,rate):
        super().__init__(name,age,eid)
        self.hrs = hrs
        self.rate = rate
    
    def Display(self):
        super().Display() #Will Search for Display() in Superclasses
        print("Working Hours: ",self.hrs,"\nRate per Hour: ",self.rate)
        print("Salary according to hrs is: ",self.hrs * self.rate)

pte = PartTimeEmployee("Almas",19,101,5,100)
pte.Display()
#e1 = Employee("Almas",19,101)
#e1.Display() #If display found in Employee Class it wil cal Display() if not then it searches for Display() in Person which is parent class of Employee 
```

# Method Overloading

```python
#Python Indirectly Supports Method Overloading through variable length arguments
def add(*no):
    ans = 0
    for i in no:
        ans = ans + i
    return ans
aa = add(20,30,-80,75.5) #enter as many args as we want!
print("The sum of entered arguments is: ",aa)
```

# Multiple Inheritance

```python
#Multiple Inheritance where third class is derived from 2 classes in this case person and employees make up third class FullTimeEmp
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def Display(self):
        print("Name: ",self.name,"\nAge: ",self.age)
class Employee:
    def __init__(self,eid):
        self.eid = eid
        
    def Display(self):
        print("E_ID: ",self.eid)
class FullTimeEmp(Person,Employee):
    def __init__(self,name,age,eid,salary):
        Person.__init__(self,name,age)
        Employee.__init__(self,eid)
        self.salary = salary
    def Display(self):
        Person.Display(self)
        Employee.Display(self)
        print("Salary is: ",self.salary)
f1 = FullTimeEmp("Almas",19,111,10000)
f1.Display()
```

```python
# Another Way to Achieve Method Overriding using multiple dispatch
from multipledispatch import dispatch

@dispatch(int,int)
def product(first,second):
    result = first * second
    print("Calling One: ",result)
@dispatch(int,int,int)
def product(first,second,third):
    result = first * second * third
    print("Calling Two: ",result)
@dispatch(float,float,float)
def product(first,second,third):
    result = first * second * third
    print("Calling Three: ",result)

product(2,2)
product(2,3,4)
product(1.5,2.5,3.5)
```
