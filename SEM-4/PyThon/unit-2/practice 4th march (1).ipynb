```python
#Program to print 10 first 10 natural numbers using for and while loop

#using for loop
print("Using For Loop")
for i in range(1,11):
    print(i)
print("\n\n")
#using While loop
x = 1
print("Using While Loop!!")
while x<=10:
    print(x)
    x+=1
```

```python
# printing even numbers in range entered by user
limit = int(input("Enter range: "))
if limit < 0:
    print("Invalid range!!")
else:
    print("Even Numbers in Entered Range are: \n")
    for i in range(0,limit+1):
        if i%2 == 0:
            print(i)
```

```python
limit = int(input("Enter range: "))
if limit < 0:
    print("Invalid range!!")
else:
    sum = 0
    print("Sum of odd Numbers in Entered Range is:")
    for i in range(0,limit+1):
        if i%2 != 0:
            sum+=i
    print(sum)
```

```python
#count no of vowels and consonants from string
#incomplete
line = input("Enter a String: ")
if line.isnumeric():
    print("Enter only Alphabets")
vowel_cnt = 0
consonant_cnt = 0
special_cnt = 0
line = line.upper()
print("Entered String: "+line)
for i in line:
    if i in ['A','E','I','O','U']:
        vowel_cnt += 1
    elif i not in['A','E','I','O','U']:
        consonant_cnt += 1
    else:
        special_cnt +=1
print("Vowels: ",vowel_cnt,"\nConsonants: ",consonant_cnt,"\nSpecials: ",special_cnt)
```

```python
#checking prime number

inp = int(input("Enter a number: "))
flag = True
for i in range(2,inp-1):
    if inp%i==0:
        flag = False
if flag == True:
    print("Prime!!")
else:
    print("Not Prime!!")
```
