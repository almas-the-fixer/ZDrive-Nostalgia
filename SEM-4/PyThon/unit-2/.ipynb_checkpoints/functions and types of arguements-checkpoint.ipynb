# Functions and their argument types!!

```python
# Function with required arguments
def arithmetic(n1,n2):
    add = n1 + n2
    sub = n1 - n2
    mul = n1 * n2
    div = n1 / n2
    
   # ans = [n1,n2,add,n1,n2,sub,n1,n2,mul,n1,n2,div]
    return n1,n2,add,n1,n2,sub,n1,n2,mul,n1,n2,div #will return tuple

aa = arithmetic(10,20) #both parameters are required else will throw error
print(aa)
```

```python
# Function with one default argument
def arithmetic2(n1,n2=10):
    add = n1 + n2
    sub = n1 - n2
    mul = n1 * n2
    div = n1 / n2
    
   # ans = [n1,n2,add,n1,n2,sub,n1,n2,mul,n1,n2,div]
    return n1,n2,add,n1,n2,sub,n1,n2,mul,n1,n2,div #will return tuple

bb = arithmetic2(10) #one parameter is required
print(bb)
```

```python
# Kayword arguments
def arithmetic3(n1,n2):
    add = n1 + n2
    sub = n1 - n2
    mul = n1 * n2
    div = n1 / n2
    
   # ans = [n1,n2,add,n1,n2,sub,n1,n2,mul,n1,n2,div]
    return n1,n2,add,n1,n2,sub,n1,n2,mul,n1,n2,div #will return tuple

cc = arithmetic(n2 = 10,n1 = 20) #specifying var values with keywords!
print(cc)
```

```python
# Variable length arguments

def add(*no):
    ans = 0
    for i in no:
        ans = ans + i
    return ans
aa = add(10,20,30,50,60,70,80,-80,75.5) #enter as many args as we want!
print("The sum of entered arguments is: ",aa)
```

```python
#Variable length keyword arguments
def kvargs(**kv):    #arguments in key value pair
    for k,v in kv.items():
        print("%s = %s"%(k,v)) 

kvargs(a="Py",b="Thon",c="keyword",d="length",e="arguments")
```
