# Taking Inputs in Lists,tuples and Dictionaries

# 1: Lists

```python
no = int(input("Enter no of values in list: "))

list = []
for i in range(0,no):
    x = int(input("Enter Value for List: "))
    list.append(x)
print(list)
```

# 2:Tuples

```python
no = int(input("Enter no of values in tuples: "))
x = []
for i in range(0,no):
    y = int(input("Enter Value for tuple: "))
    x.append(y)
    tuple1 = tuple(x)
print(tuple1)
type(tuple1)
```

```python
no = int(input("Enter no of values in dictionary: "))
dd = []
for i in range(0,no):
    x = int(input("Enter ID: "))
    y = input("Enter Name: ")
    z = input("Enter Division: ")
    d = {"ID":x,"Name":y,"Division":z}
    dd.append(d)
print(dd)
type(dd)
```
