```python
class Bank:
    def __init__(self,bank_name,branch_id,branch_name,amount):
        self.bank_name = bank_name
        self.branch_id = branch_id
        self.branch_name = branch_name
        self.amount = amount
    def Display(self):
        print("Bank Name: ",self.bank_name,"\nBranch ID: ",self.branch_id,"\nBranch Name: ",self.branch_name,"\nAmount Deposited: ",self.amount)
        
class SBI(Bank):
    def __init__(self,bank_name,branch_id,branch_name,amount,years,interest):
        super().__init__(bank_name,branch_id,branch_name,amount)
        self.years = years
        self.interest = interest
    def Display(self):
        super().Display()
        print("Period of Time: ",self.years)
    def Returns(self):
        result = (self.amount * self.interest * self.years)/100
        print("Return Of Investment in SBI is: ",result)
#incomplete from here .......
class HDFC(Bank):
    def __init__(self,bank_name,branch_id,branch_name,amount,years,interest):
        super().__init__(bank_name,branch_id,branch_name,amount)
        self.years = years
        self.interest = interest
    def Display(self):
        super().Display()
        print("Period of Time: ",self.years)

        
        
class YESBANK(Bank):
    def __init__(self,bank_name,branch_id,branch_name,years):
        super().__init__(bank_name,branch_id,branch_name)
        self.years = years

    def Display(self):
        super().Display()
        print("Period of Time: ",self.years)
        
        
s1 = SBI("SBI",101,"Juhapura Branch",100000,1,7)
s1.Display()
s1.Returns()
```
